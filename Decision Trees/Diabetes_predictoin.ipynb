```python
import numpy as mp
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_diabetes

data = load_diabetes()

print(data.feature_names)
```

```python
dataset = pd.DataFrame(data.data, columns=['age', 'sex', 'bmi', 'bp', 's1', 's2', 's3', 's4', 's5', 's6'])
```

```python
dataset.head()
```

```python
X = dataset
y = data['target']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
X_train.corr()
```

```python
import seaborn as sns
plt.figure(figsize=(15, 5))
sns.heatmap(X_train.corr(), annot=True)
```

```python
from sklearn.tree import DecisionTreeRegressor

model = DecisionTreeRegressor(
    max_depth=4,
    min_samples_split=10,
    min_samples_leaf=5
)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

print(f"The Mean Squared Error is: {mean_squared_error(y_test, y_pred)}")
print(f"The Mean Abosulute Error is: {mean_absolute_error(y_test, y_pred)}")
print(f"The R2 Score is: {r2_score(y_test, y_pred)}")
```
