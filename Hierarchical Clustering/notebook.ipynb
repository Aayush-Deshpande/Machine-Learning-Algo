```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
from sklearn.datasets import make_blobs
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import linkage, dendrogram
```

```python
warnings.filterwarnings('ignore')
```

```python
X, y = make_blobs(n_samples=300, centers=4, cluster_std=1.1, random_state=42)
df = pd.DataFrame(X, columns=['feature_1', 'feature_2'])
df.head()
```

```python
sample_df = df.sample(40, random_state=42)
linked = linkage(sample_df, method='ward')
plt.figure(figsize=(10, 5))
dendrogram(linked)
plt.title('Hierarchical Clustering Dendrogram')
plt.show()
```

```python
hierarchical = AgglomerativeClustering(n_clusters=4)
df['cluster'] = hierarchical.fit_predict(df)
df['cluster'].value_counts()
```

```python
sns.scatterplot(data=df, x='feature_1', y='feature_2', hue='cluster', palette='deep')
plt.title('Hierarchical Clustering')
plt.show()
```
