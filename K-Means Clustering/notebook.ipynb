```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
```

```python
warnings.filterwarnings('ignore')
```

```python
X, y = make_blobs(n_samples=300, centers=4, cluster_std=1.1, random_state=42)
df = pd.DataFrame(X, columns=['feature_1', 'feature_2'])
df.head()
```

```python
inertia = []
for i in range(1, 11):
    model = KMeans(n_clusters=i, random_state=42, n_init=10)
    model.fit(df)
    inertia.append(model.inertia_)

plt.plot(range(1, 11), inertia, marker='o')
plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')
plt.title('Elbow Method')
plt.show()
```

```python
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
df['cluster'] = kmeans.fit_predict(df)
kmeans.cluster_centers_
```

```python
sns.scatterplot(data=df, x='feature_1', y='feature_2', hue='cluster', palette='deep')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=200, c='black', marker='X')
plt.title('K Means Clustering')
plt.show()
```
