```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
warnings.filterwarnings('ignore')
```

```python
X, y = make_blobs(n_samples=300, centers=4, cluster_std=1.1, random_state=42)
df = pd.DataFrame(X, columns=['feature_1', 'feature_2'])
df.head()
```

```python
silhouette_scores = []
for i in range(2, 11):
    model = KMeans(n_clusters=i, random_state=42, n_init=10)
    labels = model.fit_predict(df)
    silhouette_scores.append(silhouette_score(df, labels))

silhouette_scores
```

```python
plt.plot(range(2, 11), silhouette_scores, marker='o')
plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score Evaluation')
plt.show()
```

```python
best_k = silhouette_scores.index(max(silhouette_scores)) + 2
kmeans = KMeans(n_clusters=best_k, random_state=42, n_init=10)
df['cluster'] = kmeans.fit_predict(df)
print(best_k)
print(max(silhouette_scores))
```

```python
sns.scatterplot(data=df, x='feature_1', y='feature_2', hue='cluster', palette='deep')
plt.title('Silhouette Based Clustering')
plt.show()
```
